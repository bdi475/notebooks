### 🎯 Challenge 5: Convert `"deal_date"` to a DateTime Type and Extract Inception Year

#### 👇 Tasks

- ✔️ Convert the `"deal_date"` column in the `df` DataFrame to a DateTime type (`datetime64[ns]`).
  - Hint: Use the `pd.to_datetime()` method.
- ✔️ Add a new column named `"deal_year"` to the DataFrame that extracts the year from the `"deal_date"` column.
  - Hint: Use the `.dt` accessor to extract the year from a DateTime column.
  - The `"deal_year"` column should contain integer values representing the year (e.g., `2021`, `2022`, etc.).
- ✔️ Directly update the `df` DataFrame without creating a new DataFrame.

#### 📤 Expected Output

|     | deal_date  | deal_year |
| --: | :--------- | --------: |
|   0 | 2021-12-16 |      2021 |
|   1 | 2021-10-18 |      2021 |
|   2 | 2022-01-14 |      2022 |

```
deal_date dtype: datetime64[ns]
deal_year dtype: int32
```


````python
# YOUR CODE BEGINS
df["deal_date"] = pd.to_datetime(df["deal_date"])
df["deal_year"] = df["deal_date"].dt.year
# YOUR CODE ENDS

# Select and display the relevant columns to verify the changes
display(df[["deal_date", "deal_year"]].head(3))
print(f"deal_date dtype: {df['deal_date'].dtype}")
print(f"deal_year dtype: {df['deal_year'].dtype}")
````

#### 🧭 Check Your Work

Run the code cell below to test your solution.

````python
_test_case = "extract-deal-year"
_points = 3

df_backup2 = df_backup.copy()
df_backup2["deal_date"] = df_backup2["deal_date"].astype(
    "date".upper().lower() + "TIME".lower() + "64[ns]"
)
my_dtfunc = pd.to_datetime
df_backup2["deal_year"] = my_dtfunc(df_backup2["deal_date"]) \
    .dt.year

pd.testing.assert_frame_equal(
    df[["deal_id", "deal_date", "deal_year"]]
    .sort_values("deal_id")
    .reset_index(drop=True),
    df_backup2[["deal_id", "deal_date", "deal_year"]]
    .sort_values("deal_id")
    .reset_index(drop=True),
    check_dtype=False,
    check_column_type=False,
)
````

### 🎯 Challenge 7: Total Deal Amount by Country

#### 👇 Tasks

  - The resulting DataFrame should have two columns: `"country"` and `"deal_amount"`.

### 🎯 Challenge 8: Number of Companies by Deal Type

#### 👇 Tasks

- ✔️ Group the `df` DataFrame by the `"deal_type"` column and calculate the number of companies in each deal type.
  - The resulting DataFrame should have two columns: `"deal_type"` and `"num_companies"`.
- ✔️ Store the resulting DataFrame to a new variable named `df_num_companies_by_deal_type`.
